```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#1
df = pd.read_excel("music_catalog.xlsx")
print("First 10 strings:")
print(df.head(10))
favorite_tracks = df[df["Is_Favorite"] == 1]
favorite_count = favorite_tracks.shape[0]
print("\nFavorite tracks:", favorite_count)
mean_favorite_rating = favorite_tracks["User_Rating"].mean()
print("Mean User_Rating for favorites:",
      round(mean_favorite_rating, 2))

```

```python
#2
filtered_tracks = df[
    (df["Streams_million"] >= 200) &
    (df["User_Rating"] >= 8)
]
popular_tracks = (
    row for _, row in filtered_tracks.iterrows()
    if row["Genre"] == "Pop"
)
print("\nFirst 15 Pop-tracks:")
for i, track in enumerate(popular_tracks):
    if i >= 15:
        break
    print(
        track["Title"],
        "|",
        track["Artist"],
        "|",
        track["Streams_million"],
        "|",
        track["User_Rating"]
    )
```

```python
#3

df["success_per_award"] = df.apply(
    lambda row: row["Streams_million"] /
    (row["Awards_won"] + 1),
    axis=1
)
top_success = df.sort_values(
    by="success_per_award",
    ascending=False
)
print("\nTop-10 by success_per_award:")
print(
    top_success[
        ["Title", "Artist", "success_per_award"]
    ].head(10)
)
```

```python
#4
favorite_artist_count = (
    favorite_tracks.groupby("Artist")
    .size()
    .to_dict()
)
top_artists = sorted(
    favorite_artist_count.items(),
    key=lambda x: x[1],
    reverse=True
)[:5]
print("\nTop-5 artists by favorite tracks:")
for artist, count in top_artists:
    print(artist, ":", count)
```

```python
#5
artist_genre_set = set(
    zip(
        favorite_tracks["Artist"],
        favorite_tracks["Genre"]
    )
)
print("\nCount of unique combinations Artist + Genre:")
print(len(artist_genre_set))
```

```python
#6
num_data = df[
    [
        "Streams_million",
        "User_Rating",
        "Duration_sec",
        "BPM",
        "Awards_won",
        "Awards_nominated"
    ]
].to_numpy()
favorite_streams = favorite_tracks[
    "Streams_million"
].to_numpy()
streams_mean = np.mean(favorite_streams)
streams_median = np.median(favorite_streams)
streams_std = np.std(favorite_streams)
print("\nStatistics of Streams_million for favorite tracks:")
print("Mean:", streams_mean)
print("Median:", streams_median)
print("Std:", streams_std)
max_stream_index = np.argmax(
    df["Streams_million"].to_numpy()
)
print("\nIndex of track with maximum streams:")
print(max_stream_index)
```

```python
#7
df["success_per_award"] = df.apply(
    lambda row: row["Streams_million"] /
    (row["Awards_won"] + 1),
    axis=1
)
favorite_tracks = df[df["Is_Favorite"] == 1]
rating_pivot = pd.pivot_table(
    favorite_tracks,
    index="Genre",
    columns="Country",
    values="User_Rating",
    aggfunc="mean"
)

print("\nPivot table User_Rating:")

print(rating_pivot.head())
success_pivot = pd.pivot_table(
    favorite_tracks,
    index="Genre",
    columns="Country",
    values="success_per_award",
    aggfunc="mean"
)
print("\nPivot table success_per_award:")
print(success_pivot.head())
rating_pivot.to_csv(
    "student4_rating_pivot.csv"
)
success_pivot.to_csv(
    "student4_success_pivot.csv"
)
print("\nCSV файлы успешно сохранены.")
```

```python
#8
top_streams = favorite_tracks.sort_values(
    by="Streams_million",
    ascending=False
)
print("\nТоп-10 любимых треков по стримам:")
print(
    top_streams[
        ["Title", "Artist",
         "Streams_million"]
    ].head(10)
)

top_ratings = favorite_tracks.sort_values(
    by="User_Rating",
    ascending=False
)
print("\nТоп-10 любимых треков по рейтингу:")
print(
    top_ratings[
        ["Title", "Artist",
         "User_Rating"]
    ].head(10)
)
```

```python
#9
plt.figure(figsize=(8, 5))
plt.hist(favorite_tracks["Streams_million"],
         bins=20)
plt.xlabel("Streams_million")
plt.ylabel("Количество")
plt.title("Гистограмма Streams_million")
plt.grid(True)
plt.savefig("favorite_streams_hist.png")
plt.close()
plt.figure(figsize=(10, 6))
genres = favorite_tracks["Genre"].unique()
for genre in genres:
    genre_df = favorite_tracks[
        favorite_tracks["Genre"] == genre
    ]
    plt.scatter(
        genre_df["Streams_million"],
        genre_df["User_Rating"],
        label=genre,
        alpha=0.6
    )
plt.xlabel("Streams_million")
plt.ylabel("User_Rating")
plt.title(
    "Streams_million vs User_Rating"
)
plt.grid(True)
plt.legend()
plt.savefig("streams_vs_rating.png")
plt.close()
print("\nMatplotlib графики сохранены.")
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.countplot(
    data=favorite_tracks,
    x="Genre"
)
plt.xticks(rotation=45)
plt.title(
    "Количество любимых треков по жанрам"
)
plt.savefig("favorite_genre_countplot.png")
plt.close()
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=favorite_tracks,
    x="Genre",
    y="success_per_award"
)
plt.xticks(rotation=45)
plt.title(
    "Success per Award by Genre"
)
plt.savefig("success_per_award_boxplot.png")
plt.close()
corr_columns = [
    "Streams_million",
    "User_Rating",
    "Duration_sec",
    "BPM",
    "Popularity",
    "Awards_won",
    "Awards_nominated"
]
corr_matrix = df[corr_columns].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.savefig("correlation_heatmap.png")
plt.close()
print("\nSeaborn графики сохранены.")
```
