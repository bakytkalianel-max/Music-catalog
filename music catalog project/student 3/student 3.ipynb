```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#1
df = pd.read_excel("music_catalog.xlsx")
print("First 10 strings:")
print(df.head(10))
unique_artists = df["Artist"].nunique()
unique_albums = df["Album"].nunique()
print("\nUnique Artists:", unique_artists)
print("\nUnique Albums:", unique_albums)
mean_rating = df["User_Rating"].mean()
print("Mean User_Rating:", round(mean_rating, 2))
```

```python
#2
dance_hits = df[
    (df["Energy"] >= 0.8) &
    (df["Danceability"] >= 0.7)
]
print("\nDance Hits:")
print(
    dance_hits[
        ["Title", "Artist", "Energy", "Danceability"]
    ].head(10)
)
```

```python
#3
rock_tracks = (
    row for _, row in df.iterrows()
    if row['Genre'] == "Rock"
)
print("\nFirst 15 Rock-tracks:")
for i, track in enumerate(rock_tracks):
    if i >= 15:
        break
    print(
        track["Title"],
        "|",
        track["Artist"],
        "|",
        track["Streams_million"],
        "|",
        track["User_Rating"]
    )
```

```python
high_rating = df[df["User_Rating"] >= 8]
artist_rating_count = (
    high_rating.groupby("Artist").size().to_dict()
)
top_artists = sorted(
    artist_rating_count.items(),
    key=lambda x: x[1],
    reverse=True
)[:5]
print("\nТоп-5 артистов по количеству треков с рейтингом >= 8:")
for artist, count in top_artists:
    print(artist, ":", count)
```

```python
#5
genre_country_set = set(
    zip(df["Genre"], df["Country"])
)
print("\nCount of unique combinations Genre + Country:")
print(len(genre_country_set))
```

```python
#6
df["popularity_per_stream"] = df.apply(
    lambda row: row["Popularity"] / (row["Streams_million"] + 1),
    axis=1
)
top_popularity = df.sort_values(
    by="popularity_per_stream",
    ascending=False
)
print("\nTop-10 by popularity by stream:")
print(
    top_popularity[
    ["Title", "Artist", "popularity_per_stream"]
    ].head(10)
)
```

```python
#7
numeric_array = df[
    [
        "Streams_million",
        "User_Rating",
        "Duration_sec",
        "BPM",
        "Awards_won",
        "Awards_nominated"
    ]
].to_numpy()
streams = df["Streams_million"].to_numpy()
streams_mean = np.mean(streams)
streams_median = np.median(streams)
streams_std = np.std(streams)
print("\nStreams_million statistics:")
print("Mean:", streams_mean)
print("Median:", streams_median)
print("Std:", streams_std)
max_stream_index = np.argmax(streams)
print("Индекс трека с максимальным Streams_million:")
print(max_stream_index)
```

```python
#8
rating_pivot = pd.pivot_table(
    df,
    index="Genre",
    columns="Country",
    values="User_Rating",
    aggfunc="mean"
)
popularity_pivot = pd.pivot_table(
    df,
    index="Genre",
    columns="Country",
    values="popularity_per_stream",
    aggfunc="mean"
)
rating_pivot.to_csv("student3_rating_pivot.csv")
popularity_pivot.to_csv("student3_popularity_pivot.csv")
print("\nPivot tables сохранены.")
```

```python
#9
plt.figure(figsize=(8, 5))

plt.hist(df["User_Rating"], bins=20)

plt.xlabel("User Rating")
plt.ylabel("Количество")
plt.title("Гистограмма User_Rating")
plt.grid(True)

plt.savefig("hist_user_rating.png")
plt.close()
plt.figure(figsize=(10, 6))

genres = df["Genre"].unique()

for genre in genres:
    genre_df = df[df["Genre"] == genre]

    plt.scatter(
        genre_df["Streams_million"],
        genre_df["Popularity"],
        label=genre,
        alpha=0.6
    )

plt.xlabel("Streams_million")
plt.ylabel("Popularity")
plt.title("Streams_million vs Popularity")
plt.grid(True)

plt.legend()

plt.savefig("scatter_streams_popularity.png")
plt.close()
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))

sns.countplot(data=df, x="Genre")

plt.xticks(rotation=45)
plt.title("Количество треков по жанрам")

plt.savefig("countplot_genre.png")
plt.close()
print("\nMatplotlib графики сохранены.")

```

```python
#10
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))

sns.countplot(data=df, x="Genre")

plt.xticks(rotation=45)
plt.title("Количество треков по жанрам")

plt.savefig("countplot_genre.png")
plt.close()
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Genre",
    y="popularity_per_stream"
)

plt.xticks(rotation=45)
plt.title("Popularity per Stream by Genre")

plt.savefig("boxplot_popularity_genre.png")
plt.close()
corr_columns = [
    "User_Rating",
    "Streams_million",
    "Duration_sec",
    "BPM",
    "Popularity",
    "Awards_won",
    "Awards_nominated"
]

corr_matrix = df[corr_columns].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.savefig("heatmap_correlation.png")
plt.close()

print("\nSeaborn графики сохранены.")
```
